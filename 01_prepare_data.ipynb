# Prepare Synthetic Data

This notebook generates a synthetic token stream (~50M tokens) using a simple integer tokenizer. The data is saved to `synthetic_data.pt`.

```python
import torch
import random
import math

# Parameters
num_tokens = 50_000_000
vocab_size = 50257  # typical GPT vocab size
seq_len = 1024

# Generate random token ids
tokens = torch.randint(low=0, high=vocab_size, size=(num_tokens,), dtype=torch.long)
torch.save(tokens, 'synthetic_data.pt')
print('Saved synthetic_data.pt with', tokens.numel(), 'tokens')
```
